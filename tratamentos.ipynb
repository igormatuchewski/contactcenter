## Definindo funções úteis
Em todo o script que crio costumo definir algumas funções para alterar a tipagem de colunas:

```python
## converter a coluna para string
def to_str(df, coluna):
    df[coluna] = df[coluna].astype('str')

## converter a coluna para int
def to_int(df, coluna):
    df[coluna] = df[coluna].astype('int')

## converter a coluna para float
def to_float(df, coluna):
    df[coluna] = df[coluna].astype('float')

## converter a coluna para datetime
def to_datetime(df, coluna):
    df[coluna] = pd.to_datetime(df[coluna])
```

## Gerando uma base normalizada

A base normalizada irá ser uma junção das colunas mais relevantes dentre os dataframes disponibilizados e tratados até aqui.

Primeiramente irei definir um novo dataframe que será uma cópia do df_chamadas:

```python
df_normalizada = df_chamadas.copy()
```

Agora irei verificar a quantidade de linhas e colunas atualmente no dataframe. O importante aqui neste ponto é fazer verificações na quantidade de linhas após cada merge (join) com outros dataframes, para verificar se não há duplicidades sendo incluidas no dataframe e/ou a multiplicação de algumas linhas:

```python
df_normalizada.shape
```

Verificado que há 9998 linhas irei realizar a primeira operação de merge. Irei, a partir da coluna 'id_operador', agregar as colunas 'carga_horaria', 'equipe' e 'supervisor' que estão localizadas no df_operadores:

Realizado com sucesso, agora vamos verificar se houve incremento de linhas no dataframe:

```python
df_normalizada.shape
```

Ótimo, nenhuma linha foi duplicada, o que é reflexo do bom tratamento de dados que foi feito anteriormente.
